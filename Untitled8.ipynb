```python
# ✅ 1. Subir archivo desde tu PC
from google.colab import files
uploaded = files.upload()

# ✅ 2. Cargar y explorar el dataset
import pandas as pd

df = pd.read_csv("data_balanceada.csv")
df = df.drop(columns=["Unnamed: 0", "isFlaggedFraud"])
df.head()
```

```python
# ✅ 3. Preprocesamiento: normalizar solo transacciones normales
from sklearn.preprocessing import MinMaxScaler

df_normal = df[df['isFraud'] == 0].drop(columns=["isFraud"])
scaler = MinMaxScaler()
X_normal_scaled = scaler.fit_transform(df_normal)

input_dim = X_normal_scaled.shape[1]
encoding_dim = 8
```

```python
# ✅ 4. Construcción del autoencoder
from keras.models import Model
from keras.layers import Input, Dense
from keras.optimizers import Adam

input_layer = Input(shape=(input_dim,))
encoded = Dense(16, activation='relu')(input_layer)
encoded = Dense(encoding_dim, activation='relu')(encoded)
decoded = Dense(16, activation='relu')(encoded)
decoded = Dense(input_dim, activation='sigmoid')(decoded)

autoencoder = Model(inputs=input_layer, outputs=decoded)
autoencoder.compile(optimizer=Adam(learning_rate=0.001), loss='mean_squared_error')
```

```python
# ✅ 5. Entrenamiento
history = autoencoder.fit(
    X_normal_scaled, X_normal_scaled,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    shuffle=True,
    verbose=1
)
```

```python
# ✅ 6. Detección de anomalías
import numpy as np

X_eval = df.drop(columns=["isFraud"])
y_eval = df["isFraud"].values
X_eval_scaled = scaler.transform(X_eval)

reconstructed = autoencoder.predict(X_eval_scaled)
reconstruction_error = np.mean(np.power(X_eval_scaled - reconstructed, 2), axis=1)

threshold = np.percentile(reconstruction_error[y_eval == 0], 95)
anomaly_pred = reconstruction_error > threshold
```

```python
# ✅ 7. Evaluación del modelo
from sklearn.metrics import classification_report, confusion_matrix

print(f"Umbral definido: {threshold:.5f}\n")

print("📊 Matriz de confusión:")
print(confusion_matrix(y_eval, anomaly_pred))

print("\n📈 Reporte de clasificación:")
print(classification_report(y_eval, anomaly_pred, digits=4))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Asumiendo que ya tienes anomaly_pred y y_eval definidos
conf_matrix = confusion_matrix(y_eval, anomaly_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Normal", "Fraude"], yticklabels=["Normal", "Fraude"])
plt.title("Matriz de Confusión - Autoencoder")
plt.xlabel("Predicción")
plt.ylabel("Etiqueta Verdadera")
plt.tight_layout()
plt.show()
```
